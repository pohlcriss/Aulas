```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dados = {
    "Mes": ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"],
    "Notebook": [20, 25, 22, 30, 35, 40],
    "Mouse": [50, 55, 48, 60, 65, 70],
    "Teclado": [30, 28, 35, 40, 38, 45]
}

df = pd.DataFrame(dados)
produtos = ["Notebook", "Mouse", "Teclado"]

print("Primeiras linhas:")
print(df.head())

print("\nDimensões do DataFrame:")
print(df.shape)

print("\nEstatísticas descritivas:")
print(df.describe())

print("\nIndicadores de vendas:")
for produto in produtos:
    vendas = df[produto].to_numpy()
    print(f"\n{produto}:")
    print(f"Média: {np.mean(vendas):.2f}")
    print(f"Maior venda: {np.max(vendas)}")
    print(f"Menor venda: {np.min(vendas)}")

plt.figure(figsize=(8, 5))

for produto in produtos:
    plt.plot(df["Mes"], df[produto], marker="o", label=produto)

plt.title("Evolução das vendas por mês")
plt.xlabel("Mês")
plt.ylabel("Quantidade vendida")
plt.legend()
plt.grid()
plt.show()

totais = df[produtos].sum()

plt.figure(figsize=(8, 5))
totais.plot(kind="bar")
plt.title("Total vendido por produto")
plt.xlabel("Produto")
plt.ylabel("Quantidade total")
plt.xticks(rotation=0)
plt.show()

plt.figure(figsize=(8, 5))
plt.hist(df["Mouse"], bins=5, edgecolor="black")
plt.title("Distribuição das vendas de Mouse")
plt.xlabel("Quantidade vendida")
plt.ylabel("Frequência")
plt.show()

print("\nInterpretação dos resultados:")
print(f"Produto com maior volume total: {totais.idxmax()} ({totais.max()} unidades)")

for produto in produtos:
    mesMaiorVenda = df.loc[df[produto].idxmax(), "Mes"]
    maiorVenda = df[produto].max()
    print(f"{produto}: maior venda em {mesMaiorVenda}, com {maiorVenda} unidades.")

vendasNotebook = df["Notebook"]
mediaNotebook = np.mean(vendasNotebook)

cores = ["tomato" if venda < mediaNotebook else "steelblue"
         for venda in vendasNotebook]

plt.figure(figsize=(8, 5))
plt.bar(df["Mes"], vendasNotebook, color=cores)
plt.axhline(mediaNotebook, color="black", linestyle="--",
            label=f"Média: {mediaNotebook:.2f}")
plt.title("Vendas de Notebook em relação à média")
plt.xlabel("Mês")
plt.ylabel("Quantidade vendida")
plt.legend()
plt.show()

mesesAbaixoMedia = df.loc[df["Notebook"] < mediaNotebook, "Mes"].tolist()
print(f"\nMédia de vendas de Notebook: {mediaNotebook:.2f}")
print(f"Meses abaixo da média: {', '.join(mesesAbaixoMedia)}")

vendasNotebook = df["Notebook"]
mediaNotebook = np.mean(vendasNotebook)

cores = ["red" if venda < mediaNotebook else "blue"
         for venda in vendasNotebook]

plt.figure(figsize=(8, 5))
plt.bar(df["Mes"], vendasNotebook, color=cores)
plt.axhline(mediaNotebook, color="black", linestyle="--",
            label=f"Média: {mediaNotebook:.2f}")

plt.title("Vendas de Notebook em relação à média")
plt.xlabel("Mês")
plt.ylabel("Quantidade vendida")
plt.legend()
plt.show()
```
