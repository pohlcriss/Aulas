```python

import random

historico = []

while True:
    numero_sorteado = random.randint(0, 15)
    tentativas = 0

    print("\n=== JOGO DE ADIVINHAÇÃO ===")
    print("Tente adivinhar um número de 0 a 15!")

    while True:
        palpite = int(input("Digite seu palpite: "))
        tentativas += 1

        if palpite == numero_sorteado:
            print("Parabéns! Você acertou!")
            print("Quantidade de tentativas:", tentativas)

            historico.append(
                f"Número sorteado: {numero_sorteado} | "
                f"Tentativas: {tentativas}"
            )

            with open("historico.txt", "a") as arquivo:
                arquivo.write(historico[-1] + "\n")

            break

        elif palpite < numero_sorteado:
            print("O número sorteado é maior!")

        else:
            print("O número sorteado é menor!")

    continuar = input("\nDeseja jogar novamente? (s/n): ").lower()

    if continuar != "s":
        break

print("\n=== HISTÓRICO DE JOGOS ===")

if historico:
    for jogo in historico:
        print(jogo)
else:
    print("Nenhum jogo registrado.")

print("\nHistórico salvo no arquivo historico.txt")
```

```python
import csv

lista_compras = []
total_compras = 0

print("=== LISTA DE COMPRAS ===")

while True:
    produto = input("\nDigite o nome do produto: ")

    quantidade = int(input("Digite a quantidade: "))

    valor = float(input("Digite o valor unitário do produto: "))

    subtotal = quantidade * valor

    lista_compras.append([
        produto,
        quantidade,
        valor,
        subtotal
    ])

    total_compras += subtotal

    continuar = input("\nDeseja adicionar outro produto? (s/n): ").lower()

    if continuar != "s":
        break

print("\n=== RESUMO DAS COMPRAS ===")

for item in lista_compras:
    print(
        f"Produto: {item[0]} | "
        f"Quantidade: {item[1]} | "
        f"Valor unitário: R$ {item[2]:.2f} | "
        f"Subtotal: R$ {item[3]:.2f}"
    )

print(f"\nTOTAL DAS COMPRAS: R$ {total_compras:.2f}")

with open("lista_compras.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo, delimiter=";")

    escritor.writerow([
        "Produto",
        "Quantidade",
        "Valor Unitario",
        "Subtotal"
    ])

    escritor.writerows(lista_compras)

    escritor.writerow([])
    escritor.writerow(["", "", "Total das compras", total_compras])

print("\nLista de compras salva no arquivo lista_compras.csv")
```
