```python
class Calculadora:

    def somar(self, a, b):
        return a + b

    def subtrair(self, a, b):
        return a - b

    def multiplicar(self, a, b):
        return a * b

    def dividir(self, a, b):
        if b == 0:
            return "Não é possível dividir por zero."
        return a / b

    def raiz_quadrada(self, numero):
        if numero < 0:
            return "Não existe raiz real de número negativo."

        if numero == 0:
            return 0

        # Método de Newton
        x = numero

        for i in range(20):
            x = (x + numero / x) / 2

        return x

calculadora = Calculadora()

print("=== CALCULADORA ===")
print("1 - Soma")
print("2 - Subtração")
print("3 - Multiplicação")
print("4 - Divisão")
print("5 - Raiz quadrada")

opcao = input("Escolha uma operação: ")

if opcao == "5":
    numero = float(input("Digite o número: "))
    resultado = calculadora.raiz_quadrada(numero)
else:
    numero1 = float(input("Digite o primeiro número: "))
    numero2 = float(input("Digite o segundo número: "))

    if opcao == "1":
        resultado = calculadora.somar(numero1, numero2)
    elif opcao == "2":
        resultado = calculadora.subtrair(numero1, numero2)
    elif opcao == "3":
        resultado = calculadora.multiplicar(numero1, numero2)
    elif opcao == "4":
        resultado = calculadora.dividir(numero1, numero2)
    else:
        resultado = "Opção inválida."

print("Resultado:", resultado)
```
