ATIVIDADE I

```python
import json

def gravar_personagem():
    nome = input("nome do personagem: ")
    classe = input("classe: ")
    forca = input("forca: ")
    vida = input("vida: ")
    velocidade = input("velocidade: ")

    personagem = {
        "nome": nome,
        "classe": classe,
        "forca": forca,
        "vida": vida,
        "velocidade": velocidade
    }

    with open("personagem.json", "w") as arquivo:
        json.dump(personagem, arquivo)

def ler_personagem():
    with open("personagem.json", "r") as arquivo:
        dados = json.load(arquivo)
        print(dados)

gravar_personagem()
ler_personagem()
```

ATIVIDADE II

```python
import json
import csv
import random

def gerar_numeros():
    numeros = []
    for i in range(20):
        valor = random.randint(1, 100)
        dobro = valor * 2
        dezuplo = valor * 10
        numeros.append((valor, dobro, dezuplo))
    return numeros


def salvar_txt(numeros):
    with open("numeros.txt", "w") as arquivo:
        for valor, dobro, dezuplo in numeros:
            linha = str(valor) + "," + str(dobro) + "," + str(dezuplo) + "\n"
            arquivo.write(linha)

def salvar_json(numeros):
    lista = []
    for valor, dobro, dezuplo in numeros:
        lista.append({"valor": valor, "vezes2": dobro, "vezes10": dezuplo})
    with open("numeros.json", "w") as arquivo:
        json.dump(lista, arquivo, indent=4)

def salvar_csv(numeros):
    with open("numeros.csv", "w", newline="") as arquivo:
        escritor = csv.writer(arquivo)
        escritor.writerow(["valor", "vezes2", "vezes10"])
        for valor, dobro, dezuplo in numeros:
            escritor.writerow([valor, dobro, dezuplo])

numeros = gerar_numeros()
salvar_txt(numeros)
salvar_json(numeros)
salvar_csv(numeros)
```
