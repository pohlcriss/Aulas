```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')

cidade = df[df['nome_municipio'].str.contains('Horizontina', case=False, na=False)]

pd.set_option('display.max_columns', None)

print("INFORMAÇÕES DA CIDADE:")
display(cidade)

tmin = cidade.filter(like='tmin_')

mes_menor_temp = tmin.idxmin(axis=1).iloc[0]
valor_menor_temp = tmin.min(axis=1).iloc[0]

print("\nMÊS DE MENOR TEMPERATURA")
print("Mês:", mes_menor_temp)
print("Temperatura:", valor_menor_temp, "°C")

deficit = cidade.filter(like='def_')

mes_maior_deficit = deficit.idxmax(axis=1).iloc[0]
valor_maior_deficit = deficit.max(axis=1).iloc[0]

print("\nMÊS DE MAIOR DÉFICIT HÍDRICO")
print("Mês:", mes_maior_deficit)
print("Déficit:", valor_maior_deficit, "mm")

temperaturas = cidade.filter(like='temp_media_')
verao = temperaturas[['temp_media_dez', 'temp_media_jan', 'temp_media_fev']].mean(axis=1).iloc[0]
outono = temperaturas[['temp_media_mar', 'temp_media_abr', 'temp_media_mai']].mean(axis=1).iloc[0]
inverno = temperaturas[['temp_media_jun', 'temp_media_jul', 'temp_media_ago']].mean(axis=1).iloc[0]
primavera = temperaturas[['temp_media_set', 'temp_media_out', 'temp_media_nov']].mean(axis=1).iloc[0]

print("\nMÉDIA DE TEMPERATURA POR ESTAÇÃO")
print("Verão:", verao, "°C")
print("Outono:", outono, "°C")
print("Inverno:", inverno, "°C")
print("Primavera:", primavera, "°C")
```
