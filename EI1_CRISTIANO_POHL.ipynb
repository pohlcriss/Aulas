```python
def calcularMaiorNota(notas):
    return max(notas)

def calcularMenorNota(notas):
    return min(notas)

def calcularMedia(notas):
    return sum(notas) / len(notas)

def contarAprovados(notas):
    aprovados = 0
    for nota in notas:
        if nota >= 6.0:
            aprovados += 1
    return aprovados

def contarReprovados(notas):
    reprovados = 0
    for nota in notas:
        if nota < 6.0:
            reprovados += 1
    return reprovados

def exibirRelatorio(notas):
    print("\n===== RELATÓRIO DA TURMA =====")
    print("Notas registradas :", notas)
    print("Maior nota :", calcularMaiorNota(notas))
    print("Menor nota :", calcularMenorNota(notas))
    print("Média da turma :", round(calcularMedia(notas), 2))
    print("Aprovados (>= 6.0) :", contarAprovados(notas))
    print("Reprovados (< 6.0) :", contarReprovados(notas))
    print("===============================\n")


quantidadeAlunos = int(input("Digite a quantidade de estudantes da turma: "))

listaNotas = []
for i in range(1, quantidadeAlunos + 1):
    notaAluno = float(input(f"Digite a nota do estudante {i}: "))
    listaNotas.append(notaAluno)

exibirRelatorio(listaNotas)
```

```python
def cadastrarProduto(estoque, codigo, nome, preco, quantidade):
    estoque[codigo] = {
        "nome": nome,
        "preco": preco,
        "quantidade": quantidade
    }
    return estoque

def consultarProduto(estoque, codigo):
    return estoque.get(codigo)

def calcularValorTotalEstoque(estoque):
    total = 0
    for produto in estoque.values():
        total += produto["preco"] * produto["quantidade"]
    return total

def produtoMaiorValorUnitario(estoque):
    codigoMax = max(estoque, key=lambda cod: estoque[cod]["preco"])
    return codigoMax, estoque[codigoMax]

def exibirProduto(codigo, dados):
    print("Código:", codigo, "| Nome:", dados["nome"],
          "| Preço: R$", dados["preco"], "| Quantidade:", dados["quantidade"])


# Programa principal
estoque = {}

quantidadeProdutos = int(input("Quantos produtos deseja cadastrar? "))

for i in range(quantidadeProdutos):
    print(f"\n--- Cadastro do produto {i + 1} ---")
    codigoProduto = input("Código do produto: ")
    nomeProduto = input("Nome do produto: ")
    precoProduto = float(input("Preço do produto: "))
    quantidadeProduto = int(input("Quantidade em estoque: "))
    cadastrarProduto(estoque, codigoProduto, nomeProduto, precoProduto, quantidadeProduto)

print("\n--- Todos os produtos cadastrados ---")
for codigo, dados in estoque.items():
    exibirProduto(codigo, dados)

codigoConsulta = input("\nDigite o código do produto que deseja consultar: ")
produtoEncontrado = consultarProduto(estoque, codigoConsulta)
if produtoEncontrado:
    print("\nProduto encontrado:")
    exibirProduto(codigoConsulta, produtoEncontrado)
else:
    print("Produto não encontrado.")

valorTotal = calcularValorTotalEstoque(estoque)
print("\nValor total do estoque: R$", round(valorTotal, 2))

codigoMaisCaro, dadosMaisCaro = produtoMaiorValorUnitario(estoque)
print("\nProduto de maior valor unitário:")
exibirProduto(codigoMaisCaro, dadosMaisCaro)
```
