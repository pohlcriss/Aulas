```python
def funcaoCalcularMedia(nota1, nota2):
  soma = nota1 + nota2
  media = soma / 2
  return media

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
media = funcaoCalcularMedia(nota1, nota2)
print("A média das notas é: {}".format(media))
```

```python
def maior(a, b):
  if a > b:
    return a
  else:
    return b

valor1 = int(input("Digite o primeiro número: "))
valor2 = int(input("Digite o segundo número: "))

print("O maior número é: {}".format(maior(valor1, valor2)))
```

```python
def calcula_porcento(valor, percentual=100):
  quanto = valor * percentual / 100
  return quanto

print("20% de 300 são: {:10.2f}".format(calcula_porcento(300, 20)))
print("100% de 50 são: {:10.2f}".format(calcula_porcento(50)))
```

```python
import math

n = float(input("Digite um número para calcular a raiz quadrada: "))
b = n

if n == 0:
    raiz_aproximada = 0
else:
    while True:
        p = (b + (n / b)) / 2
        quadrado = p ** 2
        diferenca = abs(n - quadrado)

        if diferenca < 0.0001:
            raiz_aproximada = p
            break

        b = p

print("Raiz quadrada aproximada pelo método de Newton:", raiz_aproximada)
print("Raiz quadrada usando SQRT:", math.sqrt(n))
```
