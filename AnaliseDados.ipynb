```python
import numpy as np

np.set_printoptions(suppress=True, precision=2)

dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98, 11700, 82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200, 95, 120.00, 2]
])

dias = np.array(["Segunda", "Terça", "Quarta", "Quinta", "Sexta", "Sábado", "Domingo"])
indicadores = np.array(["Pedidos", "Faturamento", "Clientes", "Ticket_Medio", "Reclamações"])

limFaturamento = 18000
limPedidos = 130
limClientes = 100
limAlto = 20000

print("Linhas:", dados.shape[0])
print("Colunas:", dados.shape[1])
print("Total de valores:", dados.size)
print("Dimensões:", dados.ndim)
print("Tipo dos dados:", dados.dtype)

print("2.1 Sexta-feira:", dados[4])
print("2.2 Faturamento:", dados[:, 1])
print("2.3 Reclamações:", dados[:, 4])
print("2.4 Quinta a domingo:\n", dados[3:])
print("2.5 Pedidos, faturamento e clientes:\n", dados[:, 0:3])

print("3.1 Total de pedidos:", np.sum(dados[:, 0]))
print("3.2 Faturamento total:", np.sum(dados[:, 1]))
print("3.3 Total de clientes:", np.sum(dados[:, 2]))
print("3.4 Média diária de pedidos:", np.mean(dados[:, 0]))
print("3.5 Média diária de faturamento:", np.mean(dados[:, 1]))
print("3.6 Média de reclamações por dia:", np.mean(dados[:, 4]))

print("4.1 Mais pedidos:", dias[np.argmax(dados[:, 0])], np.max(dados[:, 0]))
print("4.1 Menos pedidos:", dias[np.argmin(dados[:, 0])], np.min(dados[:, 0]))
print("4.2 Maior faturamento:", dias[np.argmax(dados[:, 1])], np.max(dados[:, 1]))
print("4.2 Menor faturamento:", dias[np.argmin(dados[:, 1])], np.min(dados[:, 1]))
print("4.3 Amplitude do faturamento:", np.max(dados[:, 1]) - np.min(dados[:, 1]))

print("5.1 Faturamento acima de", limFaturamento, ":", dias[dados[:, 1] > limFaturamento])
print("5.2 Mais de", limPedidos, "pedidos:", dias[dados[:, 0] > limPedidos])
print("5.3 Mais de", limClientes, "clientes:", dias[dados[:, 2] > limClientes])

maskPedidos150 = dados[:, 0] > 150
print("5.4 Faturamento dos dias com mais de 150 pedidos:", dados[maskPedidos150, 1])

abaixoMedia = dados[:, 1] < np.mean(dados[:, 1])
print("5.5 Dias abaixo da média semanal:", np.sum(abaixoMedia))

classificacao = np.where(dados[:, 1] >= limAlto, "Alto desempenho", "Desempenho normal")
print("6.1 Classificação:")
print(np.column_stack((dias, classificacao)))
print("6.2 Dias de alto desempenho:", np.sum(classificacao == "Alto desempenho"))

faturamento = dados[:, 1]
mediaFat = np.mean(faturamento)
medianaFat = np.median(faturamento)
desvioFat = np.std(faturamento)

print("Média:", mediaFat)
print("Mediana:", medianaFat)
print("Desvio padrão:", desvioFat)
print("Mínimo:", np.min(faturamento))
print("Máximo:", np.max(faturamento))
print("Coeficiente de variação (%):", desvioFat / mediaFat * 100)

# Sim, o faturamento varia bastante. O desvio padrão deu uns 4.551, que é
# quase 26% da média (17.714), ou seja, não é pouca coisa. O menor valor
# foi 11.700 e o maior 24.500, então o melhor dia faturou mais que o
# dobro do pior. Outra pista é que a média (17.714) ficou acima da
# mediana (16.200), porque os dias bons (quinta, sexta e sábado)
# puxam a média pra cima.

mediasCol = np.mean(dados, axis=0)
maximosCol = np.max(dados, axis=0)
minimosCol = np.min(dados, axis=0)
coefVar = np.std(dados, axis=0) / mediasCol

print("Indicadores:", indicadores)
print("8.1 Médias:", mediasCol)
print("8.2 Máximos:", maximosCol)
print("8.3 Mínimos:", minimosCol)
print("8.4 Coeficiente de variação (%):", coefVar * 100)
print("Maior variação relativa:", indicadores[np.argmax(coefVar)])

# O que mais varia é o número de reclamações (coeficiente de variação
# de uns 43%), depois vem o faturamento (uns 26%). Já o ticket médio é o
# mais estável (uns 7%), ele só anda entre 119 e 140 durante a semana.
# Usei o coeficiente de variação (desvio / média) porque os indicadores
# têm escalas muito diferentes, então comparar só o desvio padrão
# não faria sentido.

indiceMelhor = np.argmax(dados[:, 1])
print("9.1 Melhor dia:", dias[indiceMelhor])
print("9.2 Também teve mais pedidos?", indiceMelhor == np.argmax(dados[:, 0]))
print("9.3 Também teve mais reclamações?", indiceMelhor == np.argmax(dados[:, 4]))

maskIntegrada = (dados[:, 0] > 150) & (dados[:, 1] > 20000)
print("9.4 Dias com mais de 150 pedidos e faturamento acima de 20000:", np.sum(maskIntegrada))

totalPedidos = np.sum(dados[:, 0])
totalFaturamento = np.sum(dados[:, 1])
mediaPedidos = np.mean(dados[:, 0])
mediaFaturamento = np.mean(dados[:, 1])
melhorDia = dias[np.argmax(dados[:, 1])]
piorDia = dias[np.argmin(dados[:, 1])]
diaReclamacoes = dias[np.argmax(dados[:, 4])]
diasAlto = np.sum(dados[:, 1] >= limAlto)

print("RESUMO DA SEMANA")
print("Total de pedidos:", int(totalPedidos))
print("Faturamento total: R$", totalFaturamento)
print("Média diária de pedidos:", round(mediaPedidos, 2))
print("Média diária de faturamento: R$", round(mediaFaturamento, 2))
print("Melhor dia:", melhorDia)
print("Pior dia:", piorDia)
print("Maior número de reclamações:", int(np.max(dados[:, 4])), "-", diaReclamacoes)
print("Dias de alto desempenho:", diasAlto)

correlacao = np.corrcoef(dados[:, [0, 1, 4]], rowvar=False)
print("Correlação (pedidos, faturamento, reclamações):")
print(correlacao)

# CONCLUSÃO (PARTE 11):
# No geral a semana foi boa. A empresa faturou R$ 124.000 e o movimento
# foi subindo até a sexta, que foi o pico (175 pedidos e R$ 24.500).
# Quinta, sexta e sábado foram os dias de alto desempenho, e juntos
# trouxeram mais da metade do faturamento da semana.

# Dias que merecem atenção:
# - Quarta: foi o pior dia, só 98 pedidos e R$ 11.700, e ainda teve o
#   menor ticket médio (119,39). Vale pensar em alguma promoção.
# - Sexta: vendeu mais, mas também foi o dia com mais reclamações (9).
#   Pode ser que a operação não esteja dando conta do volume.
# - Quinta e sábado também tiveram bastante reclamação (7 e 6).

# Sobre a relação entre as variáveis: pedidos e faturamento andam quase
# juntos (correlação de uns 0.99), e as reclamações também sobem junto
# com eles (uns 0.94). Ou seja, quanto mais se vende, mais reclamam. Mas
# a proporção de reclamações por pedido fica entre 2% e 5%, então não
# explode. Como são só 7 dias de dados, isso é um indício e não prova
# que uma coisa causa a outra.
```
