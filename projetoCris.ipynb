```python
import requests
from datetime import date
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.float_format", "{:.2f}".format)
pd.set_option("display.max_columns", 20)
plt.rcParams["figure.figsize"] = (11, 4.5)
plt.rcParams["axes.grid"] = True

series = {"ipca": 433, "selicMes": 4390, "cambio": 3698, "desemprego": 24369}
dataInicial = "01/01/2012"
dataFinal = date.today().strftime("%d/%m/%Y")


def baixarSerie(codigo, nome, dataIni, dataFim):
    url = f"https://api.bcb.gov.br/dados/serie/bcdata.sgs.{codigo}/dados"
    parametros = {"formato": "json", "dataInicial": dataIni, "dataFinal": dataFim}
    resposta = requests.get(url, params=parametros, timeout=60)
    resposta.raise_for_status()
    return pd.DataFrame(resposta.json()).rename(columns={"valor": nome})


def prepararSerie(dfSerie, nome):
    dfSerie = dfSerie.copy()
    dfSerie["data"] = pd.to_datetime(dfSerie["data"], format="%d/%m/%Y")
    dfSerie[nome] = pd.to_numeric(dfSerie[nome], errors="coerce")
    dfSerie["data"] = dfSerie["data"].dt.to_period("M").dt.to_timestamp()
    dfSerie = dfSerie.drop_duplicates(subset="data", keep="last").sort_values("data")
    return dfSerie.set_index("data")[nome]


def detectarOutliers(serie):
    q1, q3 = serie.quantile([0.25, 0.75])
    iqr = q3 - q1
    return serie[(serie < q1 - 1.5 * iqr) | (serie > q3 + 1.5 * iqr)]


def acumular(serie):
    return ((1 + serie / 100).prod() - 1) * 100


dadosBrutos = {}
for nome, codigo in series.items():
    dadosBrutos[nome] = baixarSerie(codigo, nome, dataInicial, dataFinal)
    print(f"Serie '{nome}' (codigo {codigo}) baixada: {len(dadosBrutos[nome])} registros")

print("\n=== EXPLORACAO ===")
for nome, dfBruto in dadosBrutos.items():
    print("-" * 60)
    print(f"SERIE: {nome}")
    print(f"Registros: {dfBruto.shape[0]} | Atributos: {dfBruto.shape[1]}")
    print(dfBruto.dtypes.to_string())
    print(f"Primeira data: {dfBruto['data'].iloc[0]} | Ultima data: {dfBruto['data'].iloc[-1]}")
    print(f"Datas duplicadas: {dfBruto['data'].duplicated().sum()}")
    print(f"Valores vazios/nulos: {dfBruto[nome].isna().sum() + (dfBruto[nome] == '').sum()}")

df = pd.concat([prepararSerie(dfBruto, nome) for nome, dfBruto in dadosBrutos.items()], axis=1)
indiceCompleto = pd.date_range(df.index.min(), df.index.max(), freq="MS")
df = df.reindex(indiceCompleto)
df.index.name = "data"

print("\n=== PREPARACAO ===")
print("Formato:", df.shape)
print(df.isna().sum().to_string())
for coluna in df.columns:
    mesesNulos = df.index[df[coluna].isna()]
    if len(mesesNulos) > 0:
        print(f"{coluna}: {len(mesesNulos)} mes(es) ausente(s) de {mesesNulos.min():%m/%Y} a {mesesNulos.max():%m/%Y}")

df = df.interpolate(method="linear", limit=1, limit_area="inside")
df["ipca12m"] = ((1 + df["ipca"] / 100).rolling(12).apply(np.prod, raw=True) - 1) * 100
df["selicAa"] = ((1 + df["selicMes"] / 100) ** 12 - 1) * 100
df["juroRealMes"] = ((1 + df["selicMes"] / 100) / (1 + df["ipca"] / 100) - 1) * 100
df["juroRealAa"] = ((1 + df["selicAa"] / 100) / (1 + df["ipca12m"] / 100) - 1) * 100
df["cambioVar"] = df["cambio"].pct_change() * 100
df["ano"] = df.index.year
df["mes"] = df.index.month
display(df.tail())

for coluna in ["ipca", "cambioVar", "desemprego"]:
    outliers = detectarOutliers(df[coluna].dropna())
    print(f"{coluna}: {len(outliers)} valor(es) atipico(s)")

print("\n=== ESTATISTICAS DESCRITIVAS ===")
display(df[["ipca", "ipca12m", "selicAa", "juroRealAa", "cambio", "cambioVar", "desemprego"]].describe().T)

porAno = df.groupby("ano")["ipca"].agg(ipcaAcumulado=acumular, meses="count")
anosCompletos = porAno[porAno["meses"] == 12].copy()
display(anosCompletos)

ax = anosCompletos["ipcaAcumulado"].plot(kind="bar", color="#1f77b4")
ax.set_title("IPCA acumulado por ano (anos completos)")
ax.set_xlabel("Ano")
ax.set_ylabel("%")
for barra in ax.patches:
    ax.annotate(f"{barra.get_height():.1f}", (barra.get_x() + barra.get_width() / 2, barra.get_height()),
                ha="center", va="bottom", fontsize=8)
plt.tight_layout()
plt.show()

print("10 maiores IPCA mensais:")
display(df["ipca"].nlargest(10).to_frame("ipca (%)"))
print("5 menores IPCA mensais:")
display(df["ipca"].nsmallest(5).to_frame("ipca (%)"))

nomesMeses = ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun", "Jul", "Ago", "Set", "Out", "Nov", "Dez"]
sazonal = df.groupby("mes")["ipca"].mean()
sazonal.index = nomesMeses
ax = sazonal.plot(kind="bar", color="#ff7f0e")
ax.set_title("IPCA medio por mes do ano")
ax.set_ylabel("% ao mes")
plt.tight_layout()
plt.show()
print(f"Mes com maior inflacao media: {sazonal.idxmax()} ({sazonal.max():.2f}%)")
print(f"Mes com menor inflacao media: {sazonal.idxmin()} ({sazonal.min():.2f}%)")

fig, ax = plt.subplots()
ax.plot(df.index, df["ipca12m"], label="IPCA acumulado 12m", color="#d62728", linewidth=2)
ax.plot(df.index, df["selicAa"], label="Selic anualizada", color="#1f77b4", linewidth=2)
ax.axhline(0, color="black", linewidth=0.8)
ax.set_title("Selic e inflacao acumulada em 12 meses")
ax.set_ylabel("% ao ano")
ax.legend()
plt.tight_layout()
plt.show()

colunasCorr = ["ipca", "selicMes", "juroRealMes", "cambioVar", "desemprego"]
corr = df[colunasCorr].dropna().corr()
display(corr)
fig, ax = plt.subplots(figsize=(6.5, 5))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
ax.set_xticks(range(len(colunasCorr)))
ax.set_xticklabels(colunasCorr, rotation=45, ha="right")
ax.set_yticks(range(len(colunasCorr)))
ax.set_yticklabels(colunasCorr)
ax.grid(False)
for i in range(len(colunasCorr)):
    for j in range(len(colunasCorr)):
        ax.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center", fontsize=9)
fig.colorbar(im)
ax.set_title("Correlacao de Pearson")
plt.tight_layout()
plt.show()

baseCambio = df[["cambioVar", "ipca"]].dropna()
coefAng, coefLin = np.polyfit(baseCambio["cambioVar"], baseCambio["ipca"], 1)
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(baseCambio["cambioVar"], baseCambio["ipca"], alpha=0.6)
xReta = np.linspace(baseCambio["cambioVar"].min(), baseCambio["cambioVar"].max(), 100)
ax.plot(xReta, coefAng * xReta + coefLin, color="red", label=f"IPCA = {coefAng:.3f} * cambio + {coefLin:.2f}")
ax.set_xlabel("Variacao mensal do dolar (%)")
ax.set_ylabel("IPCA mensal (%)")
ax.set_title("Variacao cambial vs IPCA do mes")
ax.legend()
plt.tight_layout()
plt.show()

resultadosLag = {}
for lag in range(0, 13):
    par = pd.concat([df["selicAa"].shift(lag), df["ipca12m"]], axis=1).dropna()
    resultadosLag[lag] = par.corr().iloc[0, 1]
lags = pd.Series(resultadosLag, name="correlacao")
display(lags.to_frame())
ax = lags.plot(kind="bar", color="#2ca02c")
ax.set_title("Correlacao entre Selic defasada e IPCA 12m")
ax.set_xlabel("Defasagem (meses)")
ax.set_ylabel("Correlacao")
plt.tight_layout()
plt.show()
melhorLag = lags.abs().idxmax()
print(f"Maior correlacao absoluta: defasagem de {melhorLag} meses (r = {lags[melhorLag]:.2f})")

df["ipca12mFuturo"] = df["ipca12m"].shift(-12)
faixas = [-np.inf, 0, 4, 8, np.inf]
rotulos = ["Negativo (< 0%)", "Baixo (0 a 4%)", "Medio (4 a 8%)", "Alto (> 8%)"]
df["faixaJuroReal"] = pd.cut(df["juroRealAa"], bins=faixas, labels=rotulos)
resumo = (df.dropna(subset=["faixaJuroReal", "ipca12mFuturo"])
            .groupby("faixaJuroReal", observed=True)["ipca12mFuturo"]
            .agg(["count", "mean", "median"]))
resumo.columns = ["meses", "ipca12mFuturoMedia", "ipca12mFuturoMediana"]
display(resumo)
ax = resumo["ipca12mFuturoMedia"].plot(kind="bar", color="#9467bd")
ax.set_title("Inflacao dos 12 meses seguintes por faixa de juro real")
ax.set_ylabel("% acumulado")
ax.set_xlabel("")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

baseDesemprego = df.dropna(subset=["desemprego", "ipca12m"]).copy()
baseDesemprego["nivelDesemprego"] = pd.qcut(baseDesemprego["desemprego"], q=3, labels=["Baixo", "Medio", "Alto"])
display(baseDesemprego.groupby("nivelDesemprego", observed=True)
        .agg(desempregoMedio=("desemprego", "mean"),
             ipca12mMedio=("ipca12m", "mean"),
             meses=("ipca12m", "count")))

anoMax = anosCompletos["ipcaAcumulado"].idxmax()
anoMin = anosCompletos["ipcaAcumulado"].idxmin()
ultimo = df.dropna(subset=["ipca12m"]).iloc[-1]
print("\n=== CONCLUSOES ===")
print(f"Periodo analisado: {df.index.min():%m/%Y} a {df.index.max():%m/%Y} ({len(df)} meses)")
print(f"Ano com maior inflacao: {anoMax} ({anosCompletos.loc[anoMax, 'ipcaAcumulado']:.2f}%)")
print(f"Ano com menor inflacao: {anoMin} ({anosCompletos.loc[anoMin, 'ipcaAcumulado']:.2f}%)")
print(f"Mes de maior inflacao media: {sazonal.idxmax()}")
print(f"Correlacao IPCA x variacao do dolar: {corr.loc['ipca', 'cambioVar']:.2f}")
print(f"Correlacao IPCA x Selic: {corr.loc['ipca', 'selicMes']:.2f}")
print(f"Melhor defasagem Selic -> IPCA 12m: {melhorLag} meses (r = {lags[melhorLag]:.2f})")
print(f"IPCA 12m mais recente: {ultimo['ipca12m']:.2f}% em {ultimo.name:%m/%Y}")
```
